# BOW, TFIDF, Machine Learning Algorithms


```python
import pandas as pd
data = pd.read_csv('Data.txt', sep = '\t' , names = ['label', 'message'])
```

```python
data.head()
```

```python
### Data cleaning and preprocessing
import re
from nltk.stem import PorterStemmer
```

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
ps = PorterStemmer()
```

```python
corpus = []
for i in range(0, len(data)):
    review = re.sub('[^a-zA-z]',' ', data['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
    review = " ".join(review)
    corpus.append(review)
```

## Create a BOW

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 2500, ngram_range=(1,2))
```

```python
X = cv.fit_transform(corpus).toarray()
X
```

```python
y = pd.get_dummies(data['label'])
y = y.iloc[:,0].values
y
```

```python
### Train Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
X_train.shape
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_detect_model = MultinomialNB().fit(X_train, y_train)
y_pred = spam_detect_model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

## Create TF-IDF Model

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features = 2500, ngram_range=(1,2))
X = tfidf.fit_transform(corpus).toarray()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
spam_detect_model = MultinomialNB().fit(X_train, y_train)
y_pred = spam_detect_model.predict(X_test)
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

Here, we folllowed these steps:
1. Cleaning and Preprocessing
2. BOW/TF-IDF (sentences --> vectors)
3. Train test split
4. Train the model

But, the best practice is to split the data and then apply BOW/TF-IDF.
The reason for performing a train-test split before applying techniques such as Bag of Words (BoW) is primarily to prevent data leakage. Here’s a breakdown of the rationale:

1. Preventing Data Leakage: If you apply BoW or other feature engineering processes to the entire dataset before splitting, information from the test set can inadvertently influence the training data. This makes the model's evaluation not truly representative of its performance on unseen data.

2. Real-World Simulation: By splitting first, the model is trained on a dataset that simulates real-world conditions, where it encounters fresh, unprocessed data. This approach ensures that the model can generalize better to new, unseen instances.

3. Independence of Test Data: Performing the train-test split first ensures that the test set remains completely unaltered and unseen by the model during training. This is crucial for obtaining an accurate evaluation of the model's performance.

4. Best Practice: As highlighted in the course, the best practice is to perform the train-test split before applying BoW or TF-IDF, which allows you to maintain the integrity of the dataset during the training process.

This approach is essential for building robust and generalizable machine learning models.


```python
### Train Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(corpus, y, test_size = 0.20, random_state = 0)
```

```python
cv = CountVectorizer(max_features = 2500, ngram_range=(1,2))
X_train = cv.fit_transform(X_train).toarray()
X_test = cv.transform(X_test).toarray()
```

```python
spam_detect_model = MultinomialNB().fit(X_train, y_train)
y_pred = spam_detect_model.predict(X_test)
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
